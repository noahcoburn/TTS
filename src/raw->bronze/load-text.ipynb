```python
from docx import Document

def extract_clean_text(path):
    doc = Document(path)
    text = []
    for para in doc.paragraphs:
        if not is_header_or_footer(para.text):
            text.append(para.text)
    return "\n".join(text)

def is_header_or_footer(text):
    # Simple heuristic: skip page numbers or short lines
    return text.strip().isdigit() or len(text.strip()) < 5
```
